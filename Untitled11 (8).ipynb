```python
import zipfile
import os
import numpy as np
import pandas as pd
from catboost import CatBoostClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import roc_auc_score
from google.colab import files

zip_path = '/content/bank-customer-churn-prediction-dlu-course-c-5.zip'
extract_dir = '/content/bank_churn_data'

os.makedirs(extract_dir, exist_ok=True)
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)

train_df = pd.read_csv(os.path.join(extract_dir, 'train.csv'))
test_df = pd.read_csv(os.path.join(extract_dir, 'test.csv'))


def base_features(df):
    """كل الفيتشرز اللي مش محتاجة target (آمنة تتحسب مرة واحدة على الكل)."""
    out = df.copy()

    for col in ['Geography', 'Gender']:
        if col in out.columns:
            le = LabelEncoder()
            out[col] = le.fit_transform(out[col].astype(str))

    if 'Balance' in out.columns and 'EstimatedSalary' in out.columns:
        out['Balance_Salary_Ratio'] = out['Balance'] / (out['EstimatedSalary'] + 1)
        out['Zero_Balance'] = (out['Balance'] == 0).astype(int)

    if 'Age' in out.columns:
        out['Age_Group'] = pd.cut(out['Age'], bins=[0, 30, 45, 60, 100], labels=[0, 1, 2, 3]).astype(int)
        if 'NumOfProducts' in out.columns:
            out['Age_Product_Interaction'] = out['Age'] * out['NumOfProducts']
        if 'CreditScore' in out.columns:
            out['Credit_Age_Ratio'] = out['CreditScore'] / (out['Age'] + 1)

    if 'Tenure' in out.columns and 'Age' in out.columns:
        out['Tenure_Age_Ratio'] = out['Tenure'] / (out['Age'] + 1)

    if 'IsActiveMember' in out.columns and 'NumOfProducts' in out.columns:
        out['Active_Product_Interaction'] = out['IsActiveMember'] * out['NumOfProducts']

    return out


def add_surname_target_encoding(tr, va, te, target_col='Exited', smooth=10):
    """
    مهم: الـ fit بيحصل على tr بس (فولد التدريب)، مش على الداتا كلها.
    ده بيمنع الـ leakage اللي كان في الكود الأصلي.
    """
    global_mean = tr[target_col].mean()
    stats = tr.groupby('Surname')[target_col].agg(['count', 'mean'])
    stats['enc'] = (stats['count'] * stats['mean'] + smooth * global_mean) / (stats['count'] + smooth)

    tr = tr.copy(); va = va.copy(); te = te.copy()
    tr['Surname_Enc'] = tr['Surname'].map(stats['enc']).fillna(global_mean)
    va['Surname_Enc'] = va['Surname'].map(stats['enc']).fillna(global_mean)
    te['Surname_Enc'] = te['Surname'].map(stats['enc']).fillna(global_mean)
    return tr, va, te


# الفيتشرز الآمنة بس (من غير الـ Surname encoding)
train_fe = base_features(train_df)
test_fe = base_features(test_df)

drop_cols = ['id', 'CustomerId', 'Exited']
y = train_fe['Exited'].astype(int)

n_splits = 10
skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)

oof_preds = np.zeros(len(train_fe))
kaggle_test_preds = np.zeros(len(test_fe))

print(f"🚀 بدء التدريب على {n_splits}-Folds مع OOF evaluation صحيح...")

for fold, (train_idx, val_idx) in enumerate(skf.split(train_fe, y)):
    tr_fold = train_fe.iloc[train_idx].copy()
    va_fold = train_fe.iloc[val_idx].copy()

    # target encoding بيتحسب هنا بس، على تدريب الفولد ده، مفيش تسريب
    tr_fold, va_fold, te_fold = add_surname_target_encoding(tr_fold, va_fold, test_fe)

    X_tr = tr_fold.drop(columns=[c for c in drop_cols + ['Surname'] if c in tr_fold.columns])
    y_tr = tr_fold['Exited'].astype(int)
    X_va = va_fold.drop(columns=[c for c in drop_cols + ['Surname'] if c in va_fold.columns])
    y_va = va_fold['Exited'].astype(int)
    X_te = te_fold.drop(columns=[c for c in drop_cols + ['Surname'] if c in te_fold.columns])

    cat = CatBoostClassifier(
        iterations=1400, depth=6, learning_rate=0.018, l2_leaf_reg=3,
        eval_metric='AUC', verbose=0, random_seed=42 + fold
    )
    lgbm = LGBMClassifier(
        n_estimators=1200, max_depth=6, learning_rate=0.012, num_leaves=31,
        subsample=0.8, colsample_bytree=0.8, metric='auc',
        is_unbalance=True, verbose=-1, random_state=42 + fold
    )
    xgbm = XGBClassifier(
        n_estimators=1000, max_depth=5, learning_rate=0.018,
        subsample=0.8, colsample_bytree=0.8, eval_metric='auc',
        scale_pos_weight=1.8, random_state=42 + fold
    )

    cat.fit(X_tr, y_tr)
    lgbm.fit(X_tr, y_tr)
    xgbm.fit(X_tr, y_tr)

    p_cat_va = cat.predict_proba(X_va)[:, 1]
    p_lgb_va = lgbm.predict_proba(X_va)[:, 1]
    p_xgb_va = xgbm.predict_proba(X_va)[:, 1]
    fold_val_preds = (p_cat_va * 0.45) + (p_lgb_va * 0.40) + (p_xgb_va * 0.15)
    oof_preds[val_idx] = fold_val_preds

    fold_auc = roc_auc_score(y_va, fold_val_preds)
    print(f"✅ Fold {fold + 1}/{n_splits} — AUC: {fold_auc:.5f}")

    p_cat_te = cat.predict_proba(X_te)[:, 1]
    p_lgb_te = lgbm.predict_proba(X_te)[:, 1]
    p_xgb_te = xgbm.predict_proba(X_te)[:, 1]
    fold_test_preds = (p_cat_te * 0.45) + (p_lgb_te * 0.40) + (p_xgb_te * 0.15)
    kaggle_test_preds += fold_test_preds / n_splits

overall_auc = roc_auc_score(y, oof_preds)
print(f"\n📊 الـ OOF AUC الحقيقي (المتوقع على اللوحة): {overall_auc:.5f}")
print("قارنه بـ 0.93535 (أحسن نتيجة سابقة) قبل ما ترفع — لو أقل متبعتش الملف ده.")

id_col = 'id' if 'id' in test_df.columns else test_df.columns[0]
submission = pd.DataFrame({id_col: test_df[id_col], 'Exited': kaggle_test_preds})
submission.to_csv('submission_v3_fixed.csv', index=False)
files.download('submission_v3_fixed.csv')

print("\n🎉 تم إنشاء submission_v3_fixed.csv")
print(submission.head())
```

```python
import zipfile
import os
import numpy as np
import pandas as pd
from catboost import CatBoostClassifier
from xgboost import XGBClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
from google.colab import files

# ============================================================
# 🎛️ FLAGS
# ============================================================
USE_WEALTH_SCORE = True
SEEDS = [42, 123, 2024]
RUN_TAG = "exp6_stacking"
# ============================================================

zip_path = '/content/bank-customer-churn-prediction-dlu-course-c-5.zip'
extract_dir = '/content'  # زي ما ظبطناها المرة اللي فاتت

if os.path.exists(zip_path) and not os.path.exists(os.path.join(extract_dir, 'train.csv')):
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_dir)

train_df = pd.read_csv(os.path.join(extract_dir, 'train.csv'))
test_df = pd.read_csv(os.path.join(extract_dir, 'test.csv'))


def base_features(df):
    out = df.copy()
    for col in ['Geography', 'Gender']:
        if col in out.columns:
            le = LabelEncoder()
            out[col] = le.fit_transform(out[col].astype(str))
    if 'Balance' in out.columns and 'EstimatedSalary' in out.columns:
        out['Balance_Salary_Ratio'] = out['Balance'] / (out['EstimatedSalary'] + 1)
        out['Zero_Balance'] = (out['Balance'] == 0).astype(int)
        if USE_WEALTH_SCORE:
            out['Wealth_Score'] = out['Balance'] * out['EstimatedSalary']
    if 'Age' in out.columns:
        out['Age_Group'] = pd.cut(out['Age'], bins=[0, 30, 45, 60, 100], labels=[0, 1, 2, 3]).astype(int)
        if 'NumOfProducts' in out.columns:
            out['Age_Product_Interaction'] = out['Age'] * out['NumOfProducts']
        if 'CreditScore' in out.columns:
            out['Credit_Age_Ratio'] = out['CreditScore'] / (out['Age'] + 1)
    if 'Tenure' in out.columns and 'Age' in out.columns:
        out['Tenure_Age_Ratio'] = out['Tenure'] / (out['Age'] + 1)
    if 'IsActiveMember' in out.columns and 'NumOfProducts' in out.columns:
        out['Active_Product_Interaction'] = out['IsActiveMember'] * out['NumOfProducts']
    return out


def add_surname_target_encoding(tr, va, te, target_col='Exited', smooth=10):
    global_mean = tr[target_col].mean()
    stats = tr.groupby('Surname')[target_col].agg(['count', 'mean'])
    stats['enc'] = (stats['count'] * stats['mean'] + smooth * global_mean) / (stats['count'] + smooth)
    tr = tr.copy(); va = va.copy(); te = te.copy()
    tr['Surname_Enc'] = tr['Surname'].map(stats['enc']).fillna(global_mean)
    va['Surname_Enc'] = va['Surname'].map(stats['enc']).fillna(global_mean)
    te['Surname_Enc'] = te['Surname'].map(stats['enc']).fillna(global_mean)
    return tr, va, te


train_fe = base_features(train_df)
test_fe = base_features(test_df)
drop_cols = ['id', 'CustomerId', 'Exited']
y = train_fe['Exited'].astype(int)
n_splits = 10

oof_cat_all = np.zeros((len(SEEDS), len(train_fe)))
oof_xgb_all = np.zeros((len(SEEDS), len(train_fe)))
test_cat_all = np.zeros((len(SEEDS), len(test_fe)))
test_xgb_all = np.zeros((len(SEEDS), len(test_fe)))

print(f"🚀 المرحلة 1: تدريب الموديلات الأساسية (Base Models) — {RUN_TAG}\n")

for s_idx, seed in enumerate(SEEDS):
    print(f"===== Seed {seed} ({s_idx + 1}/{len(SEEDS)}) =====")
    skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=seed)

    for fold, (train_idx, val_idx) in enumerate(skf.split(train_fe, y)):
        tr_fold = train_fe.iloc[train_idx].copy()
        va_fold = train_fe.iloc[val_idx].copy()
        tr_fold, va_fold, te_fold = add_surname_target_encoding(tr_fold, va_fold, test_fe)

        X_tr = tr_fold.drop(columns=[c for c in drop_cols + ['Surname'] if c in tr_fold.columns])
        y_tr = tr_fold['Exited'].astype(int)
        X_va = va_fold.drop(columns=[c for c in drop_cols + ['Surname'] if c in va_fold.columns])
        y_va = va_fold['Exited'].astype(int)
        X_te = te_fold.drop(columns=[c for c in drop_cols + ['Surname'] if c in te_fold.columns])

        cat = CatBoostClassifier(
            iterations=3000, depth=7, learning_rate=0.02, l2_leaf_reg=5,
            bagging_temperature=0.3, border_count=128,
            eval_metric='AUC', verbose=0, random_seed=seed + fold
        )
        cat.fit(X_tr, y_tr, eval_set=(X_va, y_va), early_stopping_rounds=100, use_best_model=True)

        xgbm = XGBClassifier(
            n_estimators=3000, max_depth=5, learning_rate=0.02,
            subsample=0.8, colsample_bytree=0.8, eval_metric='auc',
            early_stopping_rounds=100, random_state=seed + fold
        )
        xgbm.fit(X_tr, y_tr, eval_set=[(X_va, y_va)], verbose=False)

        oof_cat_all[s_idx, val_idx] = cat.predict_proba(X_va)[:, 1]
        oof_xgb_all[s_idx, val_idx] = xgbm.predict_proba(X_va)[:, 1]
        test_cat_all[s_idx] += cat.predict_proba(X_te)[:, 1] / n_splits
        test_xgb_all[s_idx] += xgbm.predict_proba(X_te)[:, 1] / n_splits

        print(f"  ✅ Fold {fold + 1}/{n_splits} تم")

    print(f"  Seed {seed} تم\n")

oof_cat = oof_cat_all.mean(axis=0)
oof_xgb = oof_xgb_all.mean(axis=0)
test_cat = test_cat_all.mean(axis=0)
test_xgb = test_xgb_all.mean(axis=0)

print(f"Base CatBoost OOF AUC: {roc_auc_score(y, oof_cat):.5f}")
print(f"Base XGBoost  OOF AUC: {roc_auc_score(y, oof_xgb):.5f}")

# ============================================================
# 🥞 المرحلة 2: Stacking — Logistic Regression كـ Meta-Learner
# ============================================================
print("\n🚀 المرحلة 2: تدريب الـ Meta-Learner (Stacking)\n")

# meta-features: توقعات الموديلين + الفرق بينهم
X_meta_train = np.column_stack([oof_cat, oof_xgb, oof_cat - oof_xgb])
X_meta_test = np.column_stack([test_cat, test_xgb, test_cat - test_xgb])

# نقيّم الـ meta-learner بأمانة بـ CV منفصل (عشان الـ OOF يفضل صادق)
meta_skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=99)
oof_meta = np.zeros(len(train_fe))

for fold, (tr_idx, va_idx) in enumerate(meta_skf.split(X_meta_train, y)):
    meta_model = LogisticRegression(max_iter=1000)
    meta_model.fit(X_meta_train[tr_idx], y.iloc[tr_idx])
    oof_meta[va_idx] = meta_model.predict_proba(X_meta_train[va_idx])[:, 1]

stacking_auc = roc_auc_score(y, oof_meta)
print(f"🏆 Stacking OOF AUC: {stacking_auc:.5f}")

# مقارنة سريعة مع أحسن weighted average كنا وصلنا له
best_auc_wavg = 0
for w1 in np.arange(0, 1.01, 0.02):
    w2 = 1.0 - w1
    auc = roc_auc_score(y, w1 * oof_cat + w2 * oof_xgb)
    if auc > best_auc_wavg:
        best_auc_wavg = auc

print(f"   (مقارنة: أحسن Weighted Average كان {best_auc_wavg:.5f})")

if stacking_auc > best_auc_wavg:
    print("✅ الـ Stacking أحسن! هنستخدمه في الـ submission.")
    final_meta = LogisticRegression(max_iter=1000)
    final_meta.fit(X_meta_train, y)
    kaggle_test_preds = final_meta.predict_proba(X_meta_test)[:, 1]
    final_auc_used = stacking_auc
else:
    print("⚠️ الـ Weighted Average لسه أحسن، هنستخدمه بدل الـ Stacking.")
    best_w1 = max(np.arange(0, 1.01, 0.02), key=lambda w: roc_auc_score(y, w * oof_cat + (1 - w) * oof_xgb))
    kaggle_test_preds = best_w1 * test_cat + (1 - best_w1) * test_xgb
    final_auc_used = best_auc_wavg

print(f"\nقارن {final_auc_used:.5f} بـ 0.93585 (أحسن OOF قبل كده) و 0.93713 (أحسن نتيجة على اللوحة).")

id_col = 'id' if 'id' in test_df.columns else test_df.columns[0]
submission = pd.DataFrame({id_col: test_df[id_col], 'Exited': kaggle_test_preds})
out_name = f'submission_{RUN_TAG}.csv'
submission.to_csv(out_name, index=False)
files.download(out_name)

print(f"\n🎉 تم إنشاء {out_name}")
print(submission.head())
```

```python
import pandas as pd
import numpy as np
import xgboost as xgb
from catboost import CatBoostClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import roc_auc_score

print("🚀 جاري تحميل وتجهيز البيانات والميزات الجديدة...")

# 1. قراءة البيانات
train = pd.read_csv('/content/train.csv')
test = pd.read_csv('/content/test.csv')
test_raw = test.copy() # نسخة للاحتفاظ بعمود الـ id

# 2. تقليل التشتت (إزالة الأعمدة غير المهمة)
cols_to_drop = ['id', 'CustomerId', 'Surname']
train = train.drop(columns=cols_to_drop, errors='ignore')
test = test.drop(columns=cols_to_drop, errors='ignore')

# 3. هندسة الميزات (Feature Engineering)
for df in [train, test]:
    df['Balance_Salary_Ratio'] = df['Balance'] / df['EstimatedSalary']
    df['Tenure_to_Age'] = df['Tenure'] / df['Age']
    df['Active_CreditCard'] = df['IsActiveMember'] * df['HasCrCard']
    df['Is_Balance_Zero'] = (df['Balance'] == 0).astype(int)

# 4. تحويل النصوص لأرقام (Categorical Encoding)
le_geo = LabelEncoder()
le_gen = LabelEncoder()
train['Geography'] = le_geo.fit_transform(train['Geography'])
test['Geography'] = le_geo.transform(test['Geography'])
train['Gender'] = le_gen.fit_transform(train['Gender'])
test['Gender'] = le_gen.transform(test['Gender'])

X = train.drop(columns=['Exited'])
y = train['Exited']

# 5. إعدادات معمارية التدريب (Ensemble Setup)
SEEDS = [42, 123, 2024]
FOLDS = 10

# مصفوفات لحفظ التوقعات
oof_xgb = np.zeros(len(X))
oof_cat = np.zeros(len(X))
test_preds_xgb = np.zeros(len(test))
test_preds_cat = np.zeros(len(test))

print("\n🚀 المرحلة 1: تدريب الموديلات الأساسية (XGBoost & CatBoost) مع الميزات الجديدة")

# 6. حلقة التدريب
for seed in SEEDS:
    print(f"\n===== Seed {seed} =====")
    skf = StratifiedKFold(n_splits=FOLDS, shuffle=True, random_state=seed)

    for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
        X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
        X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]

        # --- تدريب XGBoost ---
        model_xgb = xgb.XGBClassifier(
            n_estimators=1000,
            learning_rate=0.03,
            max_depth=5,
            random_state=seed,
            eval_metric='auc',
            early_stopping_rounds=100,
            n_jobs=-1
        )
        model_xgb.fit(X_train, y_train, eval_set=[(X_val, y_val)], verbose=False)

        # حفظ توقعات XGBoost
        val_preds_xgb = model_xgb.predict_proba(X_val)[:, 1]
        oof_xgb[val_idx] += val_preds_xgb / len(SEEDS)
        test_preds_xgb += model_xgb.predict_proba(test)[:, 1] / (FOLDS * len(SEEDS))

        # --- تدريب CatBoost ---
        model_cat = CatBoostClassifier(
            iterations=1000,
            learning_rate=0.03,
            depth=6,
            random_seed=seed,
            eval_metric='AUC',
            early_stopping_rounds=100,
            verbose=False,
            thread_count=-1
        )
        model_cat.fit(X_train, y_train, eval_set=(X_val, y_val))

        # حفظ توقعات CatBoost
        val_preds_cat = model_cat.predict_proba(X_val)[:, 1]
        oof_cat[val_idx] += val_preds_cat / len(SEEDS)
        test_preds_cat += model_cat.predict_proba(test)[:, 1] / (FOLDS * len(SEEDS))

        print(f"  ✅ Fold {fold+1}/{FOLDS} تم")
    print(f"  Seed {seed} تم")

# 7. التقييم وعرض النتائج
auc_xgb = roc_auc_score(y, oof_xgb)
auc_cat = roc_auc_score(y, oof_cat)

print("\n" + "="*30)
print(f"Base XGBoost  OOF AUC: {auc_xgb:.5f}")
print(f"Base CatBoost OOF AUC: {auc_cat:.5f}")

# 8. دمج التوقعات (Weighted Average)
print("\n🚀 المرحلة 2: دمج الموديلات (Weighted Average)")
# هنا بنقسم التوقعات بالنص 50% لكل مودل لأن الاتنين أقوياء
final_oof = (0.5 * oof_xgb) + (0.5 * oof_cat)
final_auc = roc_auc_score(y, final_oof)

print(f"🏆 Weighted Average OOF AUC: {final_auc:.5f}")
print("="*30)

# 9. تجهيز ملف التقديم
final_test_preds = (0.5 * test_preds_xgb) + (0.5 * test_preds_cat)

submission = pd.DataFrame({
    'id': test_raw['id'],
    'Exited': final_test_preds
})

submission_filename = 'submission_exp7_features_blend.csv'
submission.to_csv(submission_filename, index=False)
print(f"\n🎉 تم إنشاء {submission_filename}")
```

```python
import pandas as pd
import numpy as np
from catboost import CatBoostClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import roc_auc_score
from google.colab import files

print("🚀 جاري تحميل البيانات وتطبيق هندسة الميزات المتقدمة...")

# 1. قراءة البيانات
train = pd.read_csv('/content/train.csv')
test = pd.read_csv('/content/test.csv')
test_raw = test.copy()

cols_to_drop = ['id', 'CustomerId', 'Surname']
train = train.drop(columns=cols_to_drop, errors='ignore')
test = test.drop(columns=cols_to_drop, errors='ignore')

# دمج البيانات مؤقتاً لتطبيق التجميعات (Aggregations) بدقة على كل الداتا
df_all = pd.concat([train, test], axis=0).reset_index(drop=True)

# 2. الميزات الأساسية (اللي نجحت معانا قبل كده)
df_all['Balance_Salary_Ratio'] = df_all['Balance'] / df_all['EstimatedSalary']
df_all['Tenure_to_Age'] = df_all['Tenure'] / df_all['Age']
df_all['Active_CreditCard'] = df_all['IsActiveMember'] * df_all['HasCrCard']
df_all['Is_Balance_Zero'] = (df_all['Balance'] == 0).astype(int)

# 3. الميزات المتقدمة (الجديدة)
# دمج المنطقة مع النوع
df_all['Geo_Gender'] = df_all['Geography'] + "_" + df_all['Gender']

# نصيب المنتج الواحد من الرصيد (تجنب القسمة على صفر)
df_all['Balance_per_Product'] = df_all['Balance'] / df_all['NumOfProducts'].replace(0, 1)

# تجميعات: متوسط الرصيد حسب المنطقة
geo_balance_mean = df_all.groupby('Geography')['Balance'].transform('mean')
df_all['Balance_vs_Geo_Mean'] = df_all['Balance'] - geo_balance_mean

# تجميعات: متوسط العمر حسب عدد المنتجات
prod_age_mean = df_all.groupby('NumOfProducts')['Age'].transform('mean')
df_all['Age_vs_Prod_Mean'] = df_all['Age'] - prod_age_mean

# 4. تحويل النصوص لأرقام
cat_cols = ['Geography', 'Gender', 'Geo_Gender']
for col in cat_cols:
    le = LabelEncoder()
    df_all[col] = le.fit_transform(df_all[col])

# فصل البيانات تاني لـ Train و Test
train_df = df_all[df_all['Exited'].notnull()].copy()
test_df = df_all[df_all['Exited'].isnull()].copy().drop(columns=['Exited'])

X = train_df.drop(columns=['Exited'])
y = train_df['Exited']

print("\n🚀 جاري تدريب بطلنا (CatBoost) على الميزات الجديدة...")

# 5. التدريب بـ 10 Folds عشان نضمن أقصى استقرار
FOLDS = 10
skf = StratifiedKFold(n_splits=FOLDS, shuffle=True, random_state=42)
oof_preds = np.zeros(len(X))
test_preds = np.zeros(len(test_df))

for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]

    # استخدام إعدادات CatBoost القوية
    model = CatBoostClassifier(
        iterations=1500,
        learning_rate=0.03,
        depth=6,
        eval_metric='AUC',
        random_seed=42,
        early_stopping_rounds=100,
        verbose=False,
        thread_count=-1
    )

    model.fit(X_train, y_train, eval_set=(X_val, y_val))

    val_preds = model.predict_proba(X_val)[:, 1]
    oof_preds[val_idx] = val_preds
    test_preds += model.predict_proba(test_df)[:, 1] / FOLDS

    print(f"  ✅ Fold {fold+1}/{FOLDS} تم")

# التقييم
overall_auc = roc_auc_score(y, oof_preds)
print("\n" + "="*30)
print(f"🏆 Overall OOF AUC: {overall_auc:.5f}")
print("="*30)

# تجهيز وتحميل الملف
submission = pd.DataFrame({
    'id': test_raw['id'],
    'Exited': test_preds
})
filename = 'submission_adv_features_catboost.csv'
submission.to_csv(filename, index=False)
print(f"\n🎉 تم إنشاء ملف {filename} وجاري تحميله!")
files.download(filename)
```

```python
import pandas as pd
import numpy as np
from catboost import CatBoostClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score
from google.colab import files

print("🚀 جاري تطبيق تقنية Pseudo-Labeling الخطيرة...")

# 1. تحديد التوقعات الواثقين منها جداً (أقل من 0.05 وأكبر من 0.95)
pseudo_threshold_low = 0.05
pseudo_threshold_high = 0.95

# عمل نسخة من بيانات الاختبار وتوقع المودل الأخير ليها
pseudo_test = test_df.copy()
pseudo_test['Exited_prob'] = test_preds

# فلترة الحالات الواثقين منها
confident_zeros = pseudo_test[pseudo_test['Exited_prob'] < pseudo_threshold_low].copy()
confident_zeros['Exited'] = 0

confident_ones = pseudo_test[pseudo_test['Exited_prob'] > pseudo_threshold_high].copy()
confident_ones['Exited'] = 1

# دمجهم مع بعض
pseudo_labels_df = pd.concat([confident_zeros, confident_ones]).drop(columns=['Exited_prob'])

print(f"✅ تم استخراج {len(pseudo_labels_df)} عينة شبه مؤكدة من بيانات الاختبار لإضافتها للتدريب.")

# 2. تجهيز العينات دي عشان تندمج مع بيانات التدريب
X_pseudo = pseudo_labels_df.drop(columns=['Exited'])
y_pseudo = pseudo_labels_df['Exited']

# 3. التدريب مرة تانية على (البيانات الأصلية + البيانات الجديدة الواثقين منها)
FOLDS = 10
skf = StratifiedKFold(n_splits=FOLDS, shuffle=True, random_state=42)

oof_preds_pseudo = np.zeros(len(X)) # مصفوفة التقييم (على البيانات الأصلية بس عشان التقييم يبقى دقيق)
test_preds_pseudo = np.zeros(len(test_df))

print("\n🚀 جاري إعادة تدريب بطلنا (CatBoost) على البيانات المدعومة بالـ Pseudo-Labels...")

for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
    # بيانات التدريب للـ Fold ده
    X_train_fold, y_train_fold = X.iloc[train_idx], y.iloc[train_idx]

    # دمج بيانات التدريب مع الـ Pseudo-labels (عشان المودل يتدرب عليهم سوا)
    X_train_aug = pd.concat([X_train_fold, X_pseudo], axis=0).reset_index(drop=True)
    y_train_aug = pd.concat([y_train_fold, y_pseudo], axis=0).reset_index(drop=True)

    # بيانات التحقق (مبنضفش ليها الـ Pseudo-labels عشان نختبر المودل بشفافية)
    X_val_fold, y_val_fold = X.iloc[val_idx], y.iloc[val_idx]

    model_pseudo = CatBoostClassifier(
        iterations=1500,
        learning_rate=0.03,
        depth=6,
        eval_metric='AUC',
        random_seed=42,
        early_stopping_rounds=100,
        verbose=False,
        thread_count=-1
    )

    # التدريب على البيانات المدمجة
    model_pseudo.fit(X_train_aug, y_train_aug, eval_set=(X_val_fold, y_val_fold))

    # توقع بيانات التحقق الأصلية
    oof_preds_pseudo[val_idx] = model_pseudo.predict_proba(X_val_fold)[:, 1]

    # توقع بيانات الاختبار النهائية
    test_preds_pseudo += model_pseudo.predict_proba(test_df)[:, 1] / FOLDS

    print(f"  ✅ Fold {fold+1}/{FOLDS} تم")

# 4. التقييم وعرض النتائج
pseudo_auc = roc_auc_score(y, oof_preds_pseudo)
print("\n" + "="*40)
print(f"🏆 Overall OOF AUC (with Pseudo-Labeling): {pseudo_auc:.5f}")
print("="*40)

# تجهيز وتحميل الملف
submission_pseudo = pd.DataFrame({
    'id': test_raw['id'],
    'Exited': test_preds_pseudo
})
filename_pseudo = 'submission_pseudo_labeling.csv'
submission_pseudo.to_csv(filename_pseudo, index=False)
print(f"\n🎉 تم إنشاء ملف {filename_pseudo} وجاري تحميله!")
files.download(filename_pseudo)
```

```python
!pip install lightgbm -q

import pandas as pd
import numpy as np
from catboost import CatBoostClassifier
import xgboost as xgb
import lightgbm as lgb
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import roc_auc_score
from google.colab import files

print("🚀 جاري تجهيز الميزات المتقدمة لمثلث الرعب...")

# 1. قراءة البيانات
train = pd.read_csv('/content/train.csv')
test = pd.read_csv('/content/test.csv')
test_raw = test.copy()

cols_to_drop = ['id', 'CustomerId', 'Surname']
train = train.drop(columns=cols_to_drop, errors='ignore')
test = test.drop(columns=cols_to_drop, errors='ignore')

# 2. هندسة الميزات المتقدمة
df_all = pd.concat([train, test], axis=0).reset_index(drop=True)

df_all['Balance_Salary_Ratio'] = df_all['Balance'] / df_all['EstimatedSalary']
df_all['Tenure_to_Age'] = df_all['Tenure'] / df_all['Age']
df_all['Active_CreditCard'] = df_all['IsActiveMember'] * df_all['HasCrCard']
df_all['Is_Balance_Zero'] = (df_all['Balance'] == 0).astype(int)
df_all['Geo_Gender'] = df_all['Geography'] + "_" + df_all['Gender']
df_all['Balance_per_Product'] = df_all['Balance'] / df_all['NumOfProducts'].replace(0, 1)

geo_balance_mean = df_all.groupby('Geography')['Balance'].transform('mean')
df_all['Balance_vs_Geo_Mean'] = df_all['Balance'] - geo_balance_mean

prod_age_mean = df_all.groupby('NumOfProducts')['Age'].transform('mean')
df_all['Age_vs_Prod_Mean'] = df_all['Age'] - prod_age_mean

cat_cols = ['Geography', 'Gender', 'Geo_Gender']
for col in cat_cols:
    le = LabelEncoder()
    df_all[col] = le.fit_transform(df_all[col])

train_df = df_all[df_all['Exited'].notnull()].copy()
test_df = df_all[df_all['Exited'].isnull()].copy().drop(columns=['Exited'])

X = train_df.drop(columns=['Exited'])
y = train_df['Exited']

# 3. إعدادات الموديلات
FOLDS = 10
skf = StratifiedKFold(n_splits=FOLDS, shuffle=True, random_state=42)

oof_cat = np.zeros(len(X))
test_cat = np.zeros(len(test_df))

oof_xgb = np.zeros(len(X))
test_xgb = np.zeros(len(test_df))

oof_lgb = np.zeros(len(X))
test_lgb = np.zeros(len(test_df))

print("\n🚀 جاري تدريب الموديلات الثلاثة... (الموضوع محتاج شوية صبر)")

for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]

    # -- CatBoost --
    model_cat = CatBoostClassifier(iterations=1500, learning_rate=0.03, depth=6, eval_metric='AUC', random_seed=42, early_stopping_rounds=100, verbose=False)
    model_cat.fit(X_train, y_train, eval_set=(X_val, y_val))
    oof_cat[val_idx] = model_cat.predict_proba(X_val)[:, 1]
    test_cat += model_cat.predict_proba(test_df)[:, 1] / FOLDS

    # -- XGBoost --
    model_xgb = xgb.XGBClassifier(n_estimators=1500, learning_rate=0.03, max_depth=5, eval_metric='auc', random_state=42, early_stopping_rounds=100, n_jobs=-1)
    model_xgb.fit(X_train, y_train, eval_set=[(X_val, y_val)], verbose=False)
    oof_xgb[val_idx] = model_xgb.predict_proba(X_val)[:, 1]
    test_xgb += model_xgb.predict_proba(test_df)[:, 1] / FOLDS

    # -- LightGBM --
    model_lgb = lgb.LGBMClassifier(n_estimators=1500, learning_rate=0.03, max_depth=6, random_state=42, n_jobs=-1)
    model_lgb.fit(X_train, y_train, eval_set=[(X_val, y_val)], eval_metric='auc', callbacks=[lgb.early_stopping(stopping_rounds=100, verbose=False)])
    oof_lgb[val_idx] = model_lgb.predict_proba(X_val)[:, 1]
    test_lgb += model_lgb.predict_proba(test_df)[:, 1] / FOLDS

    print(f"  ✅ Fold {fold+1}/{FOLDS} تم لجميع الموديلات")

# 4. دمج التوقعات (Weighted Average)
print("\n🚀 جاري حساب الأوزان ودمج التوقعات...")
# CatBoost 60%, XGBoost 20%, LightGBM 20%
final_oof = (0.6 * oof_cat) + (0.2 * oof_xgb) + (0.2 * oof_lgb)
final_test_preds = (0.6 * test_cat) + (0.2 * test_xgb) + (0.2 * test_lgb)

print("\n" + "="*40)
print(f"CatBoost OOF AUC: {roc_auc_score(y, oof_cat):.5f}")
print(f"XGBoost  OOF AUC: {roc_auc_score(y, oof_xgb):.5f}")
print(f"LightGBM OOF AUC: {roc_auc_score(y, oof_lgb):.5f}")
print(f"🏆 Final Ensemble OOF AUC: {roc_auc_score(y, final_oof):.5f}")
print("="*40)

# 5. حفظ الملف
submission = pd.DataFrame({'id': test_raw['id'], 'Exited': final_test_preds})
filename = 'submission_power_ensemble.csv'
submission.to_csv(filename, index=False)
print(f"\n🎉 تم إنشاء ملف {filename} وجاري تحميله!")
files.download(filename)
```

```python
import optuna
import pandas as pd
import numpy as np
from catboost import CatBoostClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score
from google.colab import files

print("🚀 جاري بدء البحث عن الخلطة السرية بـ Optuna للميزات المتقدمة...")

# تعريف دالة الهدف لـ Optuna
def objective(trial):
    params = {
        'iterations': trial.suggest_int('iterations', 1000, 2000),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.08, log=True),
        'depth': trial.suggest_int('depth', 4, 8),
        'l2_leaf_reg': trial.suggest_float('l2_leaf_reg', 2, 10, log=True),
        'eval_metric': 'AUC',
        'random_seed': 42,
        'verbose': False
    }

    # 5 أجزاء للبحث السريع
    skf_opt = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    oof_opt = np.zeros(len(X))

    for train_idx, val_idx in skf_opt.split(X, y):
        X_tr, y_tr = X.iloc[train_idx], y.iloc[train_idx]
        X_va, y_va = X.iloc[val_idx], y.iloc[val_idx]

        model = CatBoostClassifier(**params)
        model.fit(X_tr, y_tr, eval_set=[(X_va, y_va)], early_stopping_rounds=100, verbose=False)
        oof_opt[val_idx] = model.predict_proba(X_va)[:, 1]

    return roc_auc_score(y, oof_opt)

# تشغيل Optuna (15 محاولة كافية جداً)
study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=15)

print("\n" + "="*40)
print("🏆 أفضل سكور تم تحقيقه في البحث:", study.best_value)
print("✨ أفضل إعدادات لقاها:", study.best_params)
print("="*40)

print("\n🚀 جاري تدريب المودل النهائي العظيم بأفضل إعدادات...")

# تجهيز الإعدادات النهائية
best_params = study.best_params
best_params['eval_metric'] = 'AUC'
best_params['random_seed'] = 42
best_params['verbose'] = False

FOLDS = 10
skf_final = StratifiedKFold(n_splits=FOLDS, shuffle=True, random_state=42)
test_preds_final = np.zeros(len(test_df))
oof_final = np.zeros(len(X))

for fold, (train_idx, val_idx) in enumerate(skf_final.split(X, y)):
    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]

    final_model = CatBoostClassifier(**best_params)
    final_model.fit(X_train, y_train, eval_set=[(X_val, y_val)], early_stopping_rounds=100, verbose=False)

    oof_final[val_idx] = final_model.predict_proba(X_val)[:, 1]
    test_preds_final += final_model.predict_proba(test_df)[:, 1] / FOLDS
    print(f"  ✅ Final Fold {fold+1}/{FOLDS} تم")

print(f"\n🏆 Overall Final OOF AUC: {roc_auc_score(y, oof_final):.5f}")

# حفظ وتحميل الملف
submission_final = pd.DataFrame({
    'id': test_raw['id'],
    'Exited': test_preds_final
})
filename = 'submission_adv_features_optuna.csv'
submission_final.to_csv(filename, index=False)
print(f"\n🎉 تم إنشاء ملف {filename} وجاري تحميله!")
files.download(filename)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import roc_auc_score
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from google.colab import files

print("🚀 جاري بناء المعمارية العصبية (MLP)...")

# 1. تجهيز البيانات
train = pd.read_csv('/content/train.csv')
test = pd.read_csv('/content/test.csv')
test_raw = test.copy()

cols_to_drop = ['id', 'CustomerId', 'Surname']
train = train.drop(columns=cols_to_drop, errors='ignore')
test = test.drop(columns=cols_to_drop, errors='ignore')

df_all = pd.concat([train, test], axis=0).reset_index(drop=True)
df_all['Balance_Salary_Ratio'] = df_all['Balance'] / df_all['EstimatedSalary']
df_all['Tenure_to_Age'] = df_all['Tenure'] / df_all['Age']
df_all['Active_CreditCard'] = df_all['IsActiveMember'] * df_all['HasCrCard']
df_all['Is_Balance_Zero'] = (df_all['Balance'] == 0).astype(int)
df_all['Geo_Gender'] = df_all['Geography'] + "_" + df_all['Gender']
df_all['Balance_per_Product'] = df_all['Balance'] / df_all['NumOfProducts'].replace(0, 1)

geo_balance_mean = df_all.groupby('Geography')['Balance'].transform('mean')
df_all['Balance_vs_Geo_Mean'] = df_all['Balance'] - geo_balance_mean
prod_age_mean = df_all.groupby('NumOfProducts')['Age'].transform('mean')
df_all['Age_vs_Prod_Mean'] = df_all['Age'] - prod_age_mean

for col in ['Geography', 'Gender', 'Geo_Gender']:
    df_all[col] = LabelEncoder().fit_transform(df_all[col])

X = df_all[df_all['Exited'].notnull()].drop(columns=['Exited'])
y = df_all[df_all['Exited'].notnull()]['Exited'].astype(int)
test_df = df_all[df_all['Exited'].isnull()].drop(columns=['Exited'])

# 2. الـ Scaling (ضروري جداً للشبكات العصبية)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
test_df_scaled = scaler.transform(test_df)

# 3. بناء دالة المودل
def create_mlp_model(input_dim):
    model = Sequential([
        Dense(128, activation='relu', input_dim=input_dim),
        BatchNormalization(),
        Dropout(0.3),

        Dense(64, activation='relu'),
        BatchNormalization(),
        Dropout(0.2),

        Dense(32, activation='relu'),
        BatchNormalization(),
        Dropout(0.2),

        Dense(1, activation='sigmoid')
    ])
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.005),
                  loss='binary_crossentropy',
                  metrics=[tf.keras.metrics.AUC(name='auc')])
    return model

FOLDS = 5 # تقليل الطيات شوية عشان الـ NN بياخد وقت
skf = StratifiedKFold(n_splits=FOLDS, shuffle=True, random_state=42)

oof_nn = np.zeros(len(X_scaled))
test_preds_nn = np.zeros(len(test_df_scaled))

print("\n🚀 جاري تدريب الشبكة العصبية...")

for fold, (train_idx, val_idx) in enumerate(skf.split(X_scaled, y)):
    X_tr, y_tr = X_scaled[train_idx], y.iloc[train_idx]
    X_va, y_va = X_scaled[val_idx], y.iloc[val_idx]

    model = create_mlp_model(X_scaled.shape[1])

    early_stop = EarlyStopping(monitor='val_auc', mode='max', patience=15, restore_best_weights=True)
    reduce_lr = ReduceLROnPlateau(monitor='val_auc', mode='max', factor=0.5, patience=5, min_lr=1e-5)

    model.fit(X_tr, y_tr,
              validation_data=(X_va, y_va),
              epochs=100,
              batch_size=256,
              callbacks=[early_stop, reduce_lr],
              verbose=0) # قفلنا الـ logs عشان الشاشة ماتزحمش

    oof_nn[val_idx] = model.predict(X_va, batch_size=512).flatten()
    test_preds_nn += model.predict(test_df_scaled, batch_size=512).flatten() / FOLDS

final_score_nn = roc_auc_score(y, oof_nn)

print("\n" + "="*45)
print(f"🔥 OOF للشبكة العصبية (MLP): {final_score_nn:.5f}")
print("="*45)

submission = pd.DataFrame({'id': test_raw['id'], 'Exited': test_preds_nn})
filename = 'submission_keras_mlp.csv'
submission.to_csv(filename, index=False)
print(f"\n🎉 تم إنشاء ملف {filename} وجاري التحميل!")
files.download(filename)
```

```python
# ============================================================
# 🏆 الخلية النهائية: تجميع كل الموديلات + دمج محسوب + submission
# ============================================================
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score
from sklearn.linear_model import LogisticRegression
from scipy.optimize import minimize
from google.colab import files

# 1. جمّع أي موديل لقيت متغيراته موجودة في الذاكرة
# لو عندك أسامي تانية غير دي، ضيفها في اللستة زي باقي الأسطر
candidates = {
    'catboost':  ('oof_cat', 'test_preds_cat'),
    'xgboost':   ('oof_xgb', 'test_preds_xgb'),
    'lightgbm':  ('oof_lgbm', 'test_preds_lgbm'),
    'optuna':    ('oof_final', 'test_preds_final'),
    'seeds':     ('final_oof_preds', 'final_test_preds'),
    'stacking':  ('oof_filtered', 'test_preds_filtered'),
}

oof_list, test_list, model_names = [], [], []
for name, (oof_var, test_var) in candidates.items():
    if oof_var in globals() and test_var in globals():
        oof_list.append(globals()[oof_var])
        test_list.append(globals()[test_var])
        model_names.append(name)

print(f"✅ لقينا {len(oof_list)} موديلات متاحة: {model_names}")

if len(oof_list) < 2:
    print("⚠️ محتاج موديلين على الأقل عشان الدمج يكون له معنى.")
else:
    # 2. حساب أفضل أوزان دمج ممكنة (بناءً على الـ OOF)
    def neg_auc(weights):
        w = np.abs(weights) / np.sum(np.abs(weights))
        blend = sum(wi * oof for wi, oof in zip(w, oof_list))
        return -roc_auc_score(y, blend)

    result = minimize(neg_auc, np.ones(len(oof_list)) / len(oof_list), method='Nelder-Mead')
    best_weights = np.abs(result.x) / np.sum(np.abs(result.x))
    weighted_auc = -result.fun
    weighted_test_preds = sum(w * t for w, t in zip(best_weights, test_list))

    print("\n📊 أفضل أوزان دمج:")
    for name, w in zip(model_names, best_weights):
        print(f"   {name}: {w:.3f}")
    print(f"🏆 Weighted Blend OOF AUC: {weighted_auc:.5f}")

    # 3. تجربة الـ Stacking (Logistic Regression على الـ OOF)
    X_meta_train = np.column_stack(oof_list)
    X_meta_test = np.column_stack(test_list)

    meta_model = LogisticRegression()
    meta_model.fit(X_meta_train, y)
    stacking_auc = roc_auc_score(y, meta_model.predict_proba(X_meta_train)[:, 1])
    stacking_test_preds = meta_model.predict_proba(X_meta_test)[:, 1]

    print(f"🏆 Stacking OOF AUC: {stacking_auc:.5f}")

    # 4. اختيار الأفضل بينهم
    if stacking_auc > weighted_auc:
        print("\n✅ الـ Stacking كسب - هنستخدمه في التقديم")
        final_preds = stacking_test_preds
        best_auc = stacking_auc
    else:
        print("\n✅ الـ Weighted Blend كسب - هنستخدمه في التقديم")
        final_preds = weighted_test_preds
        best_auc = weighted_auc

    # 5. تجهيز ملف التقديم
    submission = pd.DataFrame({'id': test_raw['id'], 'Exited': final_preds})
    filename = f'submission_final_combo_{best_auc:.5f}.csv'
    submission.to_csv(filename, index=False)

    print(f"\n🎉 تم إنشاء {filename} (OOF AUC: {best_auc:.5f})")
    files.download(filename)
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.cluster import KMeans

# 1. قراءة البيانات من جديد
train = pd.read_csv('/content/train.csv')
test = pd.read_csv('/content/test.csv')
test_raw = test.copy()

cols_to_drop = ['id', 'CustomerId', 'Surname']
train = train.drop(columns=cols_to_drop, errors='ignore')
test = test.drop(columns=cols_to_drop, errors='ignore')

df_all = pd.concat([train, test], axis=0).reset_index(drop=True)

# 2. نفس الفيتشرز اللي كانت موجودة في test_df بتاعك
df_all['Balance_Salary_Ratio'] = df_all['Balance'] / (df_all['EstimatedSalary'] + 1e-6)
df_all['Tenure_to_Age'] = df_all['Tenure'] / df_all['Age']
df_all['Active_CreditCard'] = df_all['IsActiveMember'] * df_all['HasCrCard']

cluster_features = ['Age', 'Balance', 'EstimatedSalary']
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
df_all['Customer_Cluster'] = kmeans.fit_predict(df_all[cluster_features])

df_all['Age_Group'] = pd.qcut(df_all['Age'], q=5, labels=False)
df_all['Balance_vs_AgeGeo_Mean'] = df_all['Balance'] - df_all.groupby(['Age_Group', 'Geography'])['Balance'].transform('mean')

# 3. تحويل النصوص لأرقام (Label Encoding عادي، أسهل في التعامل مع LightGBM)
cat_cols = ['Geography', 'Gender', 'Customer_Cluster', 'Age_Group']
for col in cat_cols:
    le = LabelEncoder()
    df_all[col] = le.fit_transform(df_all[col].astype(str))

# 4. فصل train و test تاني
train_df = df_all[df_all['Exited'].notnull()].copy()
test_df = df_all[df_all['Exited'].isnull()].copy().drop(columns=['Exited'])

X = train_df.drop(columns=['Exited'])
y = train_df['Exited'].astype(int)

print("X shape:", X.shape, "| test_df shape:", test_df.shape)
print("X columns:", X.columns.tolist())
print("test_df columns:", test_df.columns.tolist())
print("\nمتطابقين؟", list(X.columns) == list(test_df.columns))
```

```python
import lightgbm as lgb
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score
import numpy as np

FOLDS = 10
skf = StratifiedKFold(n_splits=FOLDS, shuffle=True, random_state=42)
oof_lgbm_new = np.zeros(len(X))
test_preds_lgbm_new = np.zeros(len(test_df))

lgb_params = {
    'objective': 'binary', 'metric': 'auc',
    'learning_rate': 0.03, 'num_leaves': 31, 'max_depth': 5,
    'feature_fraction': 0.8, 'bagging_fraction': 0.8, 'bagging_freq': 5,
    'random_state': 42, 'verbose': -1
}

for fold, (tr_idx, val_idx) in enumerate(skf.split(X, y)):
    X_tr, y_tr = X.iloc[tr_idx], y.iloc[tr_idx]
    X_va, y_va = X.iloc[val_idx], y.iloc[val_idx]
    model = lgb.LGBMClassifier(**lgb_params, n_estimators=2000)
    model.fit(X_tr, y_tr, eval_set=[(X_va, y_va)],
              callbacks=[lgb.early_stopping(100, verbose=False)])
    oof_lgbm_new[val_idx] = model.predict_proba(X_va)[:, 1]
    test_preds_lgbm_new += model.predict_proba(test_df)[:, 1] / FOLDS
    print(f"Fold {fold+1} done")

print("\n🏆 LightGBM OOF AUC:", roc_auc_score(y, oof_lgbm_new))
```

```python
import optuna

def objective(trial):
    params = {
        'objective': 'binary', 'metric': 'auc', 'verbose': -1,
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.08, log=True),
        'num_leaves': trial.suggest_int('num_leaves', 20, 60),
        'max_depth': trial.suggest_int('max_depth', 3, 8),
        'feature_fraction': trial.suggest_float('feature_fraction', 0.6, 0.95),
        'bagging_fraction': trial.suggest_float('bagging_fraction', 0.6, 0.95),
        'bagging_freq': 5,
        'min_child_samples': trial.suggest_int('min_child_samples', 5, 50),
    }
    skf_opt = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    oof_opt = np.zeros(len(X))
    for tr_idx, val_idx in skf_opt.split(X, y):
        model = lgb.LGBMClassifier(**params, n_estimators=1500)
        model.fit(X.iloc[tr_idx], y.iloc[tr_idx],
                  eval_set=[(X.iloc[val_idx], y.iloc[val_idx])],
                  callbacks=[lgb.early_stopping(50, verbose=False)])
        oof_opt[val_idx] = model.predict_proba(X.iloc[val_idx])[:, 1]
    return roc_auc_score(y, oof_opt)

study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=20)
print("أفضل LightGBM OOF:", study.best_value)
print("أفضل إعدادات:", study.best_params)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score
from catboost import CatBoostClassifier

# نرجع نقرأ البيانات الأصلية عشان نضيف Surname للـ target encoding
train_raw = pd.read_csv('/content/train.csv')
test_raw_full = pd.read_csv('/content/test.csv')

# دالة Target Encoding بـ K-Fold (عشان نتجنب Data Leakage)
def kfold_target_encode(train_col, train_target, test_col, n_splits=5, smooth=10):
    oof_encoded = np.zeros(len(train_col))
    global_mean = train_target.mean()

    skf_te = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)
    for tr_idx, val_idx in skf_te.split(train_col, train_target):
        tr_part = train_col.iloc[tr_idx]
        target_part = train_target.iloc[tr_idx]

        stats = target_part.groupby(tr_part).agg(['mean', 'count'])
        smoothed = (stats['mean'] * stats['count'] + global_mean * smooth) / (stats['count'] + smooth)

        oof_encoded[val_idx] = train_col.iloc[val_idx].map(smoothed).fillna(global_mean)

    # للـ test: نحسب الإحصائيات من كل بيانات الـ train
    full_stats = train_target.groupby(train_col).agg(['mean', 'count'])
    full_smoothed = (full_stats['mean'] * full_stats['count'] + global_mean * smooth) / (full_stats['count'] + smooth)
    test_encoded = test_col.map(full_smoothed).fillna(global_mean)

    return oof_encoded, test_encoded

y_raw = train_raw['Exited']

# تطبيق على Surname (اللي كنا بنشيله - دلوقتي هنستغله بدل ما نرميه)
surname_oof, surname_test = kfold_target_encode(
    train_raw['Surname'], y_raw, test_raw_full['Surname'], smooth=15
)

print("✅ تم عمل Target Encoding لـ Surname")
print("مثال على القيم:", surname_oof[:5])
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score
from catboost import CatBoostClassifier

# 1. تجهيز الداتا الأصلية بنفس شكل بياناتنا
external_df_clean = external_df.drop(columns=['RowNumber', 'CustomerId'], errors='ignore')
print("شكل الداتا الخارجية بعد التنظيف:", external_df_clean.shape)

# 2. قراءة بياناتنا الأصلية تاني (train فقط - محتاجينه بعمود Surname كامل)
train_raw2 = pd.read_csv('/content/train.csv')
test_raw2 = pd.read_csv('/content/test.csv')

train_for_merge = train_raw2.drop(columns=['id', 'CustomerId'], errors='ignore')

# 3. دمج الداتا الأصلية مع بيانات المسابقة (Train فقط - Test يفضل زي ما هو)
combined_train = pd.concat([train_for_merge, external_df_clean], axis=0, ignore_index=True)
print("شكل بيانات التدريب بعد الدمج:", combined_train.shape)
print("توزيع Exited في الداتا المدموجة:")
print(combined_train['Exited'].value_counts(normalize=True))

# 4. Feature Engineering - نفس اللي عملناه قبل كده + Surname Target Encoding
def kfold_target_encode(train_col, train_target, test_col, n_splits=5, smooth=15):
    oof_encoded = np.zeros(len(train_col))
    global_mean = train_target.mean()
    skf_te = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)
    for tr_idx, val_idx in skf_te.split(train_col, train_target):
        tr_part = train_col.iloc[tr_idx]
        target_part = train_target.iloc[tr_idx]
        stats = target_part.groupby(tr_part).agg(['mean', 'count'])
        smoothed = (stats['mean'] * stats['count'] + global_mean * smooth) / (stats['count'] + smooth)
        oof_encoded[val_idx] = train_col.iloc[val_idx].map(smoothed).fillna(global_mean)
    full_stats = train_target.groupby(train_col).agg(['mean', 'count'])
    full_smoothed = (full_stats['mean'] * full_stats['count'] + global_mean * smooth) / (full_stats['count'] + smooth)
    test_encoded = test_col.map(full_smoothed).fillna(global_mean)
    return oof_encoded, test_encoded

y_combined = combined_train['Exited']
surname_oof_v2, surname_test_v2 = kfold_target_encode(
    combined_train['Surname'], y_combined, test_raw2['Surname'], smooth=15
)

combined_train['Surname_TargetEnc'] = surname_oof_v2
test_final = test_raw2.drop(columns=['id', 'CustomerId', 'Surname'], errors='ignore').copy()
test_final['Surname_TargetEnc'] = surname_test_v2.values

X_combined = combined_train.drop(columns=['Exited', 'Surname'])
y_combined = combined_train['Exited']

# 5. Label Encoding
for col in ['Geography', 'Gender']:
    le = LabelEncoder()
    X_combined[col] = le.fit_transform(X_combined[col].astype(str))
    test_final[col] = le.transform(test_final[col].astype(str))

print("\nX_combined shape:", X_combined.shape)
print("test_final shape:", test_final.shape)
print("متطابقين؟", list(X_combined.columns) == list(test_final.columns))
```

```python
best_params = {
    'iterations': 1633,
    'learning_rate': 0.026946386485504778,
    'depth': 4,
    'l2_leaf_reg': 4.59106002662027,
    'eval_metric': 'AUC',
    'verbose': False
}

FOLDS = 10
skf = StratifiedKFold(n_splits=FOLDS, shuffle=True, random_state=42)
oof_combined = np.zeros(len(X_combined))
test_preds_combined = np.zeros(len(test_final))

for fold, (tr_idx, val_idx) in enumerate(skf.split(X_combined, y_combined)):
    X_tr, y_tr = X_combined.iloc[tr_idx], y_combined.iloc[tr_idx]
    X_va, y_va = X_combined.iloc[val_idx], y_combined.iloc[val_idx]
    model = CatBoostClassifier(**best_params, random_seed=42)
    model.fit(X_tr, y_tr, eval_set=[(X_va, y_va)], early_stopping_rounds=100, verbose=False)
    oof_combined[val_idx] = model.predict_proba(X_va)[:, 1]
    test_preds_combined += model.predict_proba(test_final)[:, 1] / FOLDS
    print(f"Fold {fold+1} done")

score_combined = roc_auc_score(y_combined, oof_combined)
print(f"\n🏆 OOF AUC مع البيانات المدمجة (25k): {score_combined:.5f}")
```

```python
# نحسب الـ OOF بس على الصفوف اللي جايه من بيانات المسابقة الأصلية (أول 15000 صف)
original_data_mask = np.arange(len(X_combined)) < 15000
score_on_original_only = roc_auc_score(
    y_combined[original_data_mask],
    oof_combined[original_data_mask]
)
print(f"🎯 OOF على بيانات المسابقة الأصلية بس: {score_on_original_only:.5f}")
print("مقارنة بأفضل نتيجة سابقة (0.93842):", "تحسن ✅" if score_on_original_only > 0.93842 else "لم يتحسن ❌")
```

```python
seeds = [42, 123, 2024]
oof_combined_avg = np.zeros(len(X_combined))
test_combined_avg = np.zeros(len(test_final))

for seed in seeds:
    skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=seed)
    oof_seed = np.zeros(len(X_combined))
    test_seed = np.zeros(len(test_final))
    for tr_idx, val_idx in skf.split(X_combined, y_combined):
        model = CatBoostClassifier(**best_params, random_seed=seed)
        model.fit(X_combined.iloc[tr_idx], y_combined.iloc[tr_idx],
                  eval_set=[(X_combined.iloc[val_idx], y_combined.iloc[val_idx])],
                  early_stopping_rounds=100, verbose=False)
        oof_seed[val_idx] = model.predict_proba(X_combined.iloc[val_idx])[:, 1]
        test_seed += model.predict_proba(test_final)[:, 1] / 10
    oof_combined_avg += oof_seed / len(seeds)
    test_combined_avg += test_seed / len(seeds)

    score_orig_only = roc_auc_score(
        y_combined[original_data_mask],
        oof_combined_avg[original_data_mask]
    )
    print(f"Seed {seed} → OOF على بيانات المسابقة الأصلية بس: {score_orig_only:.5f}")
```
